```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
data2026 = pd.read_csv("2026Data.csv")
data2025 = pd.read_csv("2025Data.csv")
```

```python
new=pd.concat([data2026,data2025])
new.describe()
```

```python
new = new.rename(columns={'Total Bulk and Bags': 'vol','ASP Current Year': 'asp'})
new['yr'] = pd.to_datetime(new['Current Year Week Ending']).dt.year
new['rev'] = new.vol * new.asp
# Total U.S., same weeks (1-32) in both years
t = new[(new.Geography == 'Total U.S.') & (new.Period <= 32)]

avg_vol = t.pivot_table(index='Type', columns='yr', values='vol', aggfunc='mean')
avg_vol['% change'] = (avg_vol[2026] / avg_vol[2025] - 1) * 100  ✓
avg_vol.round(1)  ✓
```

```python
g = t.groupby(['Type', 'yr']).agg(rev=('rev', 'sum'), vol=('vol', 'sum'), simple=('asp', 'mean'))
g['weighted'] = g.rev / g.vol  ✓
```

```python
w = g['weighted'].unstack('yr') 
w['% change'] = (w[2026] / w[2025] - 1) * 100  ✓
w.round(4)
```
